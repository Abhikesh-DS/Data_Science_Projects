# Project Title: Personal Finance Tracker

```python
# Finance Tracking Program

income = 0
expenses = 0
balance = 0

while True:
    print("--- Finance Tracker ---")
    print("1. Add Income")
    print("2. Add Expense")
    print("3. View Summary")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        amount = float(input("Enter income amount: "))
        
        if amount > 0:
            income += amount
            balance += amount
            print(f"₹{amount} added to income.")
        else:
            print("Invalid amount! Income must be positive.")

    elif choice == "2":
        amount = float(input("Enter expense amount: "))
        
        if amount > 0:
            if amount > balance:
                print("Insufficient balance!")
            else:
                expenses += amount
                balance -= amount
                print(f"₹{amount} added to expenses.")
        else:
            print("Invalid amount! Expense must be positive.")

    elif choice == "3":
        print("--- Financial Summary ---")
        print(f"Total Income   : ₹{income}")
        print(f"Total Expenses : ₹{expenses}")
        print(f"Balance        : ₹{balance}")

        if balance < 500:
            print("Warning: Your savings are low!")

    elif choice == "4":
        print("Exiting Finance Tracker...")
        break

    else:
        print("Invalid choice! Please enter 1-4.")
```
